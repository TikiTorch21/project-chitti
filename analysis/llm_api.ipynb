```python
def generate_answer(question: str, context: str):
    prompt = f"""
    Use ONLY the information in the context. If it's not there, say 'I don't know'.

    Context:
    {context}

    Question:
    {question}
    """
    response = requests.post(
        "http://localhost:11434/api/chat",
        json={
            "model": "gemma3:1b",
            "messages": [{"role": "user", "content": prompt}],
            "stream": False
        }
    )
    return response.json()["message"]["content"]
```

```python
generate_answer("What is your favorite food?", "I love Chocolate, Banana, Fish. I hate apples and oranges.")
```

```python
def generate_answer_conversational(): 
    question = ""
    while question.lower() != "stop":
        question = input("What is your question? ")
        prompt = f"""
        You are a helpful intern that answers my questions.

        Question:
        {question}
        """
        response = requests.post(
            "http://localhost:11434/api/chat",
            json={
                "model": "gemma3:1b",
                "messages": [{"role": "user", "content": prompt}],
                "stream": False
            }
        )
        print(response.json()["message"]["content"])
```

```python
generate_answer_conversational()
```

```python
def chat_with_gemma(model: str = "gemma3:1b"):
    """
    A continuous chat loop using Gemma via Ollama.
    Remembers conversation in this session only.
    Type 'exit' or 'quit' to stop.
    """
    messages = []

    print("💬 Chat started with Gemma (type 'exit' to quit):\n")
    while True:
        user_input = input("You: ")
        if user_input.lower() in ("exit", "quit"):
            print("Ending chat. Goodbye!")
            break

        # Append user's message to history
        messages.append({"role": "user", "content": user_input})

        # Send request with full message history
        response = requests.post(
            "http://localhost:11434/api/chat",
            json={
                "model": model,
                "messages": messages,
                "stream": False
            }
        )
        reply = response.json()["message"]["content"]

        print(f"Gemma: {reply}\n")

        # Append assistant reply to history so it remembers context
        messages.append({"role": "assistant", "content": reply})
    
    return messages
```

```python
chat_with_gemma()
```
