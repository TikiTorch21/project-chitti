```python
def fixed_tokens(text, chunk_size=100, overlap=20):
    step = chunk_size - overlap
    chunks = []

    for i in range(0, len(text), step):
        chunk = text[i:i+chunk_size]   # slice by characters
        chunks.append(chunk)

        if i + chunk_size >= len(text):
            break

    return chunks
```

```python
text = "Hello, my name is Prateek; " * 20
fixed_tokens(text=text)
```

```python
def recursive_char_split(text, chunk_size=200, overlap=20):
    separators = ["\n\n", ". ", " ", ""]
    chunks = [text]

    for sep in separators:
        new_chunks = []
        for c in chunks:
            if len(c) <= chunk_size:
                new_chunks.append(c)
            elif sep:
                parts, cur = [], ""
                for p in c.split(sep):
                    piece = p + sep
                    if len(cur) + len(piece) <= chunk_size:
                        cur += piece
                    else:
                        if cur: new_chunks.append(cur)
                        cur = piece
                if cur: new_chunks.append(cur)
            else:  
                step = chunk_size - overlap
                for i in range(0, len(c), step):
                    new_chunks.append(c[i:i+chunk_size])
        chunks = new_chunks
    return chunks
```

```python
sample = "This is a long text. It has multiple sentences. And even paragraphs.\n\nHere is another one."
for i, c in enumerate(recursive_char_split(sample, 40, 10), 1):
    print(f"\nChunk {i}:\n{c}")
```
